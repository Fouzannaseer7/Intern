```python
3. Which of these will raise an error?
A) my_tuple = (10, 20)
B) my_tuple[0] = 30
C) my_list = [10, 20]
D) my_list[0] = 30
```

```python
✅ B) my_tuple[0] = 30
Tuples are immutable, so you cannot change their items.
```

```python
4. Which of the following is a correct way to define a tuple?
A) my_tuple = [1, 2, 3]
B) my_tuple = (1, 2, 3)
C) my_tuple = {1, 2, 3}
D) my_tuple = "1, 2, 3"
```

```python
✅ B) my_tuple = (1, 2, 3)
Round brackets ( ) are used for tuples.
```

 Dictionary Data Type in Python

```python
--A dictionary is a collection of key-value pairs used to store data
  in a structured way.
```

```python
🔹Creation:
--It means making or defining a data structure for the first time.
```

```python
student = {
    "name": "Jabbar",
    "age": 21,
    "course": "AI & DS"
}

#"name", "age", "course" → keys
#"Jbbar", 21, "AI & DS"  → values
```

```python
🔹Properties:
>Unordered (in older Python versions)
In Python 3.7+, dictionaries maintain insertion order.

>Mutable
You can change, add, or remove key-value pairs.

>Keys must be unique
Values can repeat, but keys can't.

>Can store different data types
Keys are usually strings or numbers; values can be anything.

```

```python
#Unordered:
my_dict = {"a": 1, "b": 2, "c": 3}
print(my_dict) 
```

```python
#Mutable:
person = {"name": "Ali"}
person["age"] = 25         # Add
person["name"] = "Jabbar"  # Update
del person["age"]          # Delete
print(person)
```

```python
#Keys must be unique:
d = {"a": 1, "b": 2, "a": 3}
print(d) 
```

```python
#Can store different data types:
info = {
    "name": "Sara",
    "marks": [85, 90, 92],
    "details": {"height": 160, "weight": 55}
}
print(info)
```
