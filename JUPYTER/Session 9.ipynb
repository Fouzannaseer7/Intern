```python
List Comprehension:
--It is a short way to create a new list from an existing one using a
  single line of code.
```

```python
[statment for var in iterable if cond]

--var – each item from the iterable (like a list or range)
--condition – a filter to include only certain items
--statement – what you want to do with each item
```

```python
evens = [x for x in range(10) if x % 2 == 0]

x is the variable
range(10) is the iterable (numbers from 0 to 9)
if x % 2 == 0 is the condition (only even numbers)
Final list: [0, 2, 4, 6, 8]
```

```python
🔷Tuple Data Type
--A tuple is a collection of ordered and immutable items.
```

```python
my_tuple = (10, 20, 30, 40)
print(my_tuple)
```

```python
🔹Properties of Tuples:
--Ordered: Items have a defined order
--Immutable: Cannot be changed after creation
--Allows duplicates
--Can store different data types
```

```python
1.Ordered:
--The elements in a tuple have a fixed order.
--This means each item has an index, and that order won’t change.
```

```python
t = (10, 20, 30)
print(t[0]) 
```

```python
2.Immutable:
--You cannot change, add, or remove items in a tuple after it is created.
```

```python
t = (1, 2, 3)
t[0] = 10  # ❌ Error: 'tuple' object does not support item assignment
```

```python
3.Allows Duplicates:
Tuples can have repeated values.
```

```python
t = (1, 2, 2, 3)
print(t)  
```

```python
4.Can Store Different Data Types:
--Tuples can hold a mix of integers, strings, floats, booleans, etc.
```

```python
t = (1, "hello", 3.14, True)
print(t) 
```

```python
🔹Indexing:
--Access elements using their position.
```

```python
print(my_tuple)
print(my_tuple[0])   
print(my_tuple[-1])  
```

```python
🔹Slicing:
--Get a part of the tuple.
```

```python
print(my_tuple)
print(my_tuple[1:3])   
print(my_tuple[:2])   
```

```python
🔹Tuple Methods:
--Tuples have limited methods since they are immutable.

Method	            Description
count(x)	Returns the number of times x appears
index(x)	Returns the index of first occurrence of x
```

```python
t = (1, 2, 3, 2)
print(t.count(3))  
print(t.index(2))   
```
