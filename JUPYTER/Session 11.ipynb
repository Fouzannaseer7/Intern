```python
🔶 Set Data Type in Python:
A set is a collection that is:

>Unordered
>Unindexed
>Does not allow duplicates
```

```python
my_set = {1, 2, 3, 4}
print(my_set)
```

```python
🔹Properties of Sets:
>Unordered: Items don’t follow any order.
>Unique Items: No duplicates allowed.
>Mutable: You can add or remove items.
>Cannot Access by Index.
```

```python
#Unordered:
s = {10, 20, 30}
print(s) # (order can vary)
```

```python
#Unique Items:
s = {1, 2, 2, 3}
print(s) 
```

```python
#Mutable:
s = {1, 2}
s.add(3)       # Add item
s.remove(1)    # Remove item
print(s)       
```

```python
#Cannot Access by Index:
s = {10, 20, 30}
print(s[0])    # ❌ Error: 'set' object is not subscriptable
```

```python
>Explanation:
The error means: You can't use indexing ([0]) on a set.
Sets are unordered, so Python doesn't assign positions (like index 0, 1, 2) to the elements.
That's why you can't access elements by index
```

```python
✅ Conditional Statements in Python:
--Conditional statements are used to make decisions in your code.
-- They let you run different blocks of code based on whether a condition is True or False.
```

```python
1.if Statement:
--Runs a block of code only if the condition is true.
```

```python
x = 10
if x > 5:

    print("x is greater than 5")
```

```python
2. if-else Statement:
--Runs one block if the condition is true, another if it is false.
```

```python
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is 5 or less")
```

```python
3. if-elif-else Statement:
--Checks multiple conditions in order.
```

```python
x = 0
if x > 0:
    print("Positive")
elif x == 0:
    print("Zero")
else:
    print("Negative")
```

```python
🔹Comparison Operators Used:
Operator	 Meaning	        Example
==      	Equal to	        x == 5
!=	        Not equal to	    x != 5
>	        Greater than	    x > 5
<	        Less than	        x < 5
>=	        Greater or equal	x >= 5
<=	        Less or equal	    x <= 5
```
