```python
LIST DATA TYPE:
--A list is an ordered, mutable (changeable), and iterable collection of items.
--Lists can hold elements of different data types (integers, strings, etc.).
--We are using [] bracket
```

```python
my_list = [10, "hello", 3.14, True]
print(my_list)
```

```python
#Properties of Lists:
-Ordered: Items have a defined order.
-Mutable: You can change, add, or remove elements.
-Allows Duplicates: Lists can have repeated values.
-Heterogeneous: Can store mixed data types.
```

```python
# 🔹 Creating a list
my_list = [10, "apple", 3.14, True, "apple"]
```

```python
# 🔸 Ordered: Items are stored in the same order
print("Ordered list:", my_list)
```

```python
# 🔸 Mutable: We can change, add, or remove items
my_list[1] = "banana"      # Changing an element
my_list.append("new item") # Adding an element
my_list.remove(True)       # Removing an element

print("After modifying (Mutable):", my_list)
```

```python
# 🔸 Allows Duplicates: 'apple' appears more than once
print("Duplicates allowed:", my_list)
```

```python

# 🔸 Heterogeneous: List contains different data types
print("Mixed types in list:")
for item in my_list:
    print(f"{item} ({type(item)})")
```

```python
#Indexing
-Used to access individual items by position.
>Index starts from 0 (left to right)
>Negative index -1 starts from right (last item)
```

```python
my_list = ['a', 'b', 'c', 'd']
print(my_list[0])   
print(my_list[1])  
print(my_list[2])  
print(my_list[3])  
print(my_list[-1])  
print(my_list[4])  
```

```python
#Slicing
-Used to get a part of the list.
```

```python
list[start : stop : step]
>start: starting index (included)
>stop: ending index (excluded)
>step: interval

default values

a[::]
start is 0
stop is len(str)
step is 1


a[::-1]
start : -1
stop : -len(str) - 1
    
a[::1]
start : 0
stop : len(str)
```

```python
my_list = [0, 1, 2, 3, 4, 5]
print(my_list[1:4])    # [1, 2, 3]
print(my_list[:3])     # [0, 1, 2]
print(my_list[::2])    # [0, 2, 4]
```

```python
a = 'python class'
print(a[-6::3])
```

```python
#METHODS
Common List Methods:
Method	                 Description
append(x)	    Adds x to the end of the list
insert(i, x)	Inserts x at index i
extend(list2)	Adds all elements from another list
remove(x)	    Removes the first x
pop(i)	        Removes and returns item at index i
index(x)	    Finds index of first x
count(x)	    Counts how many times x occurs
sort()	        Sorts the list (ascending)
reverse()	    Reverses the list
clear()	        Removes all items
```

```python
1.append(x):
-Adds an element x to the end of the list.
```

```python
lst = [1, 2]
lst.append(3)
print(lst)  # [1, 2, 3]
```

```python
2.insert(index, x):
-Inserts x at the specified index.
```

```python
lst = [1, 3]
lst.insert(1, 2)
print(lst)  # [1, 2, 3]
```

```python
3. extend(iterable):
-Adds all elements from another list (or any iterable).
```

```python
lst = [1, 2]
lst.extend([3, 4])
print(lst)  # [1, 2, 3, 4]
```

```python
4. remove(x):
-Removes the first occurrence of x.
```

```python
lst = [1, 2, 3, 2]
lst.remove(2)
lst.remove(3)
print(lst)  
```

```python
5. pop(index=-1):
-Removes and returns the element at index (default is last).
```

```python
lst = [1, 2, 3]
x = lst.pop()
print(x)     
print(lst)   
```

```python
6. index(x):
-Returns the index of the first occurrence of x.
```

```python
lst = ['a', 'b', 'c']
print(lst.index('b'))  
```

```python
7. count(x):
-Returns the number of times x appears.
```

```python
lst = [1, 2, 2, 3]
print(lst.count(2)) 
print(lst.count(5)) 
```

```python
8. sort():
-Sorts the list in ascending order.
```

```python
lst = [3, 1, 2]
lst.sort()
print(lst) 
```

```python
9. reverse():
Reverses the list in place
```

```python
lst = [1, 2, 3]
lst.reverse()
print(lst) 
```

```python
10. copy():
-Returns a shallow copy of the list.
```

```python
lst1 = [1, 2, 3]
lst2 = lst1.copy()
print(lst2) 
```

```python
11. clear():
-Removes all items from the list.
```

```python
lst = [1, 2, 3]
lst.clear()
print(lst) 
```
